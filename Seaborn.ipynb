# Seaborn -Data Visualization

Seaborn is a Python Visualization library built on top of Matplotlib. It provides a high level interface for statistical graphics and wors specially well with Pandas DataFrames.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Built in Datasets

There are built in datasets available in seaborn library

```python
sns.get_dataset_names()
```

Loading Dataset

```python
df=sns.load_dataset('tips')
```

```python
df
```

## Scatter plot

Scatter plot is a type of graph used to show the relationship between two numerical variable. Each dots represent one observation/data points

```python
sns.scatterplot(data=df,x='total_bill',y='tip',color='teal',linewidth=1,edgecolor='violet')
plt.show()
```

If the dots generally move upward from left to right, it indiates a positive relationship : larger bills tends to be associated with larger tips.

##Hue

Hue maps another variable to color. Most commonly, it is used with categorical variable to compare groups, but it can also represent numerical values.

```python
sns.scatterplot(data=df,x="total_bill",y="tip",hue="time")
plt.show()
```

##Style

Style shows different markers based on the column variables

```python
sns.scatterplot(data=df,x="total_bill",y="tip",hue="time",style='sex')
plt.show()
```
