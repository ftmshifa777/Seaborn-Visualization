# Seaborn -Data Visualization

Seaborn is a Python Visualization library built on top of Matplotlib. It provides a high level interface for statistical graphics and wors specially well with Pandas DataFrames.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Built in Datasets

There are built in datasets available in seaborn library

```python
sns.get_dataset_names()
```

Loading Dataset

```python
df=sns.load_dataset('tips')
```

```python
df
```

## Scatter plot

Scatter plot is a type of graph used to show the relationship between two numerical variable. Each dots represent one observation/data points

```python
sns.scatterplot(data=df,x='total_bill',y='tip',color='teal',linewidth=1,edgecolor='violet')
plt.show()
```

If the dots generally move upward from left to right, it indiates a positive relationship : larger bills tends to be associated with larger tips.

##Hue

Hue maps another variable to color. Most commonly, it is used with categorical variable to compare groups, but it can also represent numerical values.

```python
sns.scatterplot(data=df,x="total_bill",y="tip",hue="time")
plt.show()
```

##Style

Style shows different markers based on the column variables

```python
sns.scatterplot(data=df,x="total_bill",y="tip",hue="time",style='sex')
plt.show()
```

##Line Plot

```python
sns.lineplot(data=df,x="size",y="tip")
plt.show()
```

##Bar Plot

```python
sns.barplot(data=df,x='day',y='total_bill',hue='time')
plt.show()
```

##Count Plot


```python
sns.countplot(data=df, x='day',color='orange')
plt.show()
```

##Box Plot

```python
sns.boxplot(data=df,x='day',y='total_bill')
plt.show()
```

A box plot (also called a box-and-whisker plot) is a visualisation used to understand the distribution of numerical data. It is especially useful for seeing:


Median

Lower and upper quartiles

Spread of the data

Possible outliers

Comparison between groups


A box plot is a very common in EDA(Exploratory Data Analysis ) and Machine Learning.

https://media.geeksforgeeks.org/wp-content/uploads/20251121105754642527/419171619.webp






##Q1 - First Quartile

Q1 is the value below which approximately 25% of the observations fall.

Q1 forms the left/bottom edge of the box, depending on whether the box plot is horizontal or vertical.

##Median

median is the middle value of the data.

It divides the data into two halves:

In a boxplot, the median is shown as a line inside the box

##Q3 - Third Quartile

Q3 is the value below which approximately 75% of the observations fall.

##IQR - InterQuartile Range

IQR=Q3-Q1

##Whiskers

The lines extending from the box are called whiskers.

They show the range of the data that is conidered non oultlier data.

A common rule is:

$$ Lower\ Bound = Q1 - 1.5(IQR) $$ $$ Upper\ Bound = Q3 + 1.5(IQR) $$

Values outside these boundaries are typically considered potential outliers.

##Violin Plot

A violin plot combines distribution density information with a compact summary of the data.

```python
sns.violinplot(data=df,x='day',y='total_bill')
plt.show()
```

* width of the violin indicates the densit of observations at different numerical values.A wider section  means more observations are concentrated around that value.

* Narrow section: Indicates fewer observations around that value.

* Height: Represents the range of numerical values in the category

*Shape: Helps identify the distribution, including clusters and possible multiple peaks.

When to use a Violin plot when you want to:

1. Compare numerical distributions across multiple cateories.
2. Identify where values are concentrated.
3. Understand more about distribution than a simple box plot may reveal.


## Histogram

```python
sns.histplot(data=df,x='total_bill',color='blue')
plt.show()
```

## KDE Plot

KDE means Kernel Density Estimate. It provides a smooth estimate of numerical distribution.

```python
sns.kdeplot(data=df,x='total_bill')

plt.show()
```

## Histogram + KDE

A histogram + KDE plot combines both a histogram and a kernel density estimate curve in a single graph. It helps us understand the distribution of numerical data using both bars and a smooth curve

```python
sns.histplot(data=df,x='total_bill',kde=True)

plt.show()
```

## Strip Plot

A Strip plot is a statistical visualization that displays individual data points as dots across different categories. It helps us understand how numerical values are distributed within each category.

```python
sns.stripplot(data=df,x='day',y='total_bill',color='orange')
plt.show()
```

## Swarm Plot

A Swarm Plot is a statistical visualization that displays individual data points as dots across different categories. It arranges the dots so that they do not overlap, making it easier to see individual observations and where values are concentrated.


```python
sns.swarmplot(data=df,x='day',y='total_bill',color='teal')

plt.show()
```
