# Data visualization with Seaborn on Iris Dataset

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset("iris")
```

```python
df.head()
```

```python
df.columns
```

```python
df.describe()
```

```python
sns.scatterplot(data=df,x='sepal_length',y='petal_length',color='Magenta',edgecolor="blue",linewidth=2)
plt.show()
```

```python

sns.scatterplot(data=df,x='sepal_length',y='petal_length',hue='species',edgecolor='lightgreen',linewidth=2)
plt.show()
```

```python
sns.scatterplot(data=df,x='sepal_length',y='petal_length',style='species',hue='petal_width')
plt.show()
```

```python
sns.lineplot(data=df,x='sepal_length',y='petal_length',color='violet')
plt.show()
```

```python
plt.figure(figsize=(12,12))
sns.barplot(data=df,x='sepal_length',y='petal_length',hue='sepal_length')
plt.show()
```

```python

sns.countplot(data=df,x='petal_length',hue='species')
plt.xticks(rotation=90)
plt.show()
```

```python
sns.boxplot(data=df,x='species',y='petal_length',color='purple')
plt.show()
```

```python
sns.violinplot(data=df,x='species',y='sepal_length',color='red')
plt.show()
```

```python
sns.histplot(data=df,x='sepal_length',color='violet')
plt.show()
```

```python
sns.kdeplot(data=df,x='petal_length',color='red')
plt.show()
```

```python
sns.histplot(data=df,x='petal_length',color='red',kde=True)
plt.show()
```

```python
sns.stripplot(data=df,x='species',y='petal_length',color='teal')
plt.show()
#
```

```python
sns.swarmplot(data=df,x='species',y='petal_length',color='violet')
plt.show()
```
